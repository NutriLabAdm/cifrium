### 01. Сравнительный анализ тектов для определения качества транскрибации для разных моделей 

для моделей транскрибации whisper ['tiny', 'base', 'small', 'medium', 'large', 'large-v1', 'large-v2', 'large-v3']

приведен анализ для текстов Цифриум (качество звука позволяет практически любым моделям транскрибировать с небольшим количеством ошибок) и лекций по Философии с низким качеством звука 

```python
from difflib import SequenceMatcher

def highlight_diffs(line1, line2):
    sm = SequenceMatcher(None, line1, line2)
    result = ''
    for tag, i1, i2, j1, j2 in sm.get_opcodes():
        if tag == 'equal':
            result += line1[i1:i2]
        elif tag == 'delete':
            result += f'{ps.BLRED}{line1[i1:i2]}{ps._}'
        elif tag == 'insert':
            result += f'{ps.BLGREEN}{line2[j1:j2]}{ps._}'
        elif tag == 'replace':
            result += f'{ps.BLRED}{line1[i1:i2]}{ps._}{ps.BLGREEN}{line2[j1:j2]}{ps._}'
    return result


def compare_texts_precise(text1, text2):
    lines1 = text1.strip().splitlines()
    lines2 = text2.strip().splitlines()

    sm = SequenceMatcher(None, lines1, lines2)
    for tag, i1, i2, j1, j2 in sm.get_opcodes():
        if tag == 'equal':
            continue
        elif tag == 'delete':
            for line in lines1[i1:i2]:
                print(f"{ps.LRED}- {line}{ps._}")
        elif tag == 'insert':
            for line in lines2[j1:j2]:
                print(f"{ps.LGREEN}+ {line}{ps._}")
        elif tag == 'replace':
            for l1, l2 in zip(lines1[i1:i2], lines2[j1:j2]):
                print(f"~ {highlight_diffs(l1, l2)}")


# Пример использования:
text_old = """NutriChat is an intelligent assistant for personalized nutrition.
It helps users to track meals, get tips, and stay healthy."""

text_new = """NutriChat is an AI assistant for personalized nutrition and wellness.
It helps users track their meals, receive smart tips, and stay healthy."""

compare_texts_precise(text_old, text_new)
```

```python
def gradient_color(value, text):
    """
    Подсвечивает текст цветом в зависимости от значения от -10 до 10.
    -10 — красный, 0 — жёлтый, 10 — зелёный.
    """
    # Ограничим значение диапазоном [-10, 10]
    value = max(min(value, 10), -10)

    # Преобразуем значение к [0, 1] (для градиента)
    normalized = (value + 10) / 20

    # Расчёт RGB: от (255, 0, 0) -> (255, 255, 0) -> (0, 255, 0)
    if normalized < 0.5:
        # Красный → Жёлтый
        r = 255
        g = int(255 * (normalized * 2))
        b = 0
    else:
        # Жёлтый → Зелёный
        r = int(255 * (1 - (normalized - 0.5) * 2))
        g = 255
        b = 0

    # ANSI-код 24-bit цвета
    return f"\033[38;2;{r};{g};{b}m{text}\033[0m"


def gradient_bg(value, text):
    """
    Подсвечивает фон текста цветом:
    - от -10 (красный) → 0 (белый) → +10 (синий)
    """
    value = max(min(value, 10), -10)  # ограничение
    r = g = b = 255  # старт с белого

    if value < 0:
        # от -10 до 0 → красный к белому
        factor = (value + 10) / 10  # от 0 до 1
        r = 255
        g = int(255 * factor)
        b = int(255 * factor)
    elif value > 0:
        # от 0 до 10 → белый к синему
        factor = (10 - value) / 10  # от 1 до 0
        r = 100 +  int(155 * factor)
        g = 180 + int(75 * factor)
        b = 255

    return f"\033[48;2;{r};{g};{b}m{text}\033[0m"



for i in range(-10, 11):
    print(gradient_bg(i, f" {i:+7} "), end=" ")
print()

for i in range(-10, 11):
    print(gradient_color(i, f"{i:+3} → цветовой градиент"))
```

```python
import os
import re
from difflib import SequenceMatcher
from collections import Counter
import glob


def highlight_diffs(line1, line2):
    sm = SequenceMatcher(None, line1, line2)
    result = ''
    for tag, i1, i2, j1, j2 in sm.get_opcodes():
        if tag == 'equal':
            result += line1[i1:i2]
        elif tag == 'delete':
            result += f'{ps.BLRED}{line1[i1:i2]}{ps._}'
        elif tag == 'insert':
            result += f'{ps.BLGREEN}{line2[j1:j2]}{ps._}'
        elif tag == 'replace':
            result += f'{ps.BLRED}{line1[i1:i2]}{ps._}{ps.BLGREEN}{line2[j1:j2]}{ps._}'
    return result

def compare_texts_precise(text1, text2):
    lines1 = text1.strip().splitlines()
    lines2 = text2.strip().splitlines()
    sm = SequenceMatcher(None, lines1, lines2)
    for tag, i1, i2, j1, j2 in sm.get_opcodes():
        if tag == 'equal':
            continue
        elif tag == 'delete':
            for line in lines1[i1:i2]:
                print(f"{ps.LRED}- {line}{ps._}")
        elif tag == 'insert':
            for line in lines2[j1:j2]:
                print(f"{ps.LGREEN}+ {line}{ps._}")
        elif tag == 'replace':
            for l1, l2 in zip(lines1[i1:i2], lines2[j1:j2]):
                print(f"~ {highlight_diffs(l1, l2)}")

def tokenize(text):
    return re.findall(r'\b\w+\b', text.lower())

def unique_word_diff(ref_words, comp_words):
    ref_set, comp_set = set(ref_words), set(comp_words)
    sym_diff = ref_set.symmetric_difference(comp_set)
    union = ref_set.union(comp_set)
    return round(len(sym_diff) / len(union), 4) if union else 0.0

def unique_long_word_diff(ref_words, comp_words):
    ref_set = {w for w in ref_words}
    comp_set = {w for w in comp_words}
    sym_diff = ref_set.symmetric_difference(comp_set)
    union = ref_set.union(comp_set)
    return round(len(sym_diff) / len(union), 4) if union else 0.0

def top_words(words, n=300):
    return Counter(words).most_common(n)

def show_legend():
    print(f"{ps.BOLD}Легенда различий:{ps._}")
    print(f"{ps.BLRED}- Удалено (есть в эталоне, нет в сравнении){ps._}")
    print(f"{ps.BLGREEN}+ Добавлено (есть в сравнении, нет в эталоне){ps._}")
    print(f"{ps.BLRED}красным{ps._} и {ps.BLGREEN}зелёным{ps._} подсвечиваются замены внутри строк\n")

def display_top_diff(ref_top, comp_top, ref_words):
    # Сортируем слова по длине
    sorted_words = sorted(comp_top, key=lambda x: len(x[0]), reverse=True)
    
    # Заголовок таблицы для 6 колонок
    header = " | ".join([f"{'word':<18} | {'ref':<3} | {'cmp':<3} | {'Δ':^4}"] * 8)
    print(f"{ps.BOLD}{header}{ps._}")
    print("-" * len(header))  # 34 символа на столбец × 6

    row = []
    count = 0
    for word, comp_count in sorted_words:
        if len(word) <= 1:
            continue
        ref_count = ref_words.get(word, 0)
        delta = comp_count - ref_count

        _delta_c = gradient_bg(delta, f" {delta:+3} ")
        
        row.append(f"{word:<18} | {ref_count:<3} | {comp_count:<3} |{_delta_c}")
        count += 1

        if count % 8 == 0:
            print(" | ".join(row))
            row = []

    # Выводим остатки, если есть
    if row:
        while len(row) < 6:
            row.append(" " * 36)
        print(" | ".join(row))


def compare_text(ref_text = None, comp_text=None):
    ref_words = tokenize(ref_text)
    ref_size = len(ref_text)
    ref_word_count = Counter(ref_words)

    show_legend()
       
    if ref_text == comp_text: 
        print(f"\n{ps.BLYELLOW}=== Одинаковый текст === {ps._}") 
        return

    comp_words = tokenize(comp_text)
    comp_size = len(comp_text)
    comp_top = top_words(comp_words)

    matcher = SequenceMatcher(None, ref_text, comp_text)
    similarity = matcher.ratio()
    diff_percent = round((1 - similarity) * 100, 2)
    size_diff_percent = round((comp_size - ref_size) / ref_size * 100, 2)
    word_diff_ratio = unique_word_diff(ref_words, comp_words)
    long_word_diff_ratio = unique_long_word_diff(ref_words, comp_words)

    print(f"{ps.GRAY}Коэффициент совпадения (символы): {similarity:.4f} ({100 - diff_percent:.2f}%)")
    print(f"Размер эталона: {ref_size} симв. | Сравниваемый: {comp_size} симв. | Δ: {size_diff_percent:+.2f}%")
    print(f"Уникальные слова, различие: {word_diff_ratio*100:.2f}%")
    print(f"Уникальные длинные слова (длиной > 3 символа), различие: {long_word_diff_ratio*100:.2f}%")
    
    display_top_diff(ref_word_count, comp_top, ref_word_count)
    print()
    compare_texts_precise(ref_text, comp_text)



def load_text(path):
    with open(path, encoding='utf-8') as f:
        return f.read()

# Запуск

REFERENCE_NAME = r"lectures_2025\Общественное движение в России 2четверь XIX в_01\transcripts\Общественное движение в России 2четверь XIX в_01_large.txt"
FOLDER_MASK = r"lectures_2025\Общественное движение в России 2четверь XIX в_01\transcripts\*.txt"

reference_text = load_text(REFERENCE_NAME)

for filename in glob.glob(FOLDER_MASK):
    print(f"\n{ps.BLBLUE}=== Сравнение с {filename} ==={ps._}")
    
    compare_text(reference_text, load_text(filename))
```
