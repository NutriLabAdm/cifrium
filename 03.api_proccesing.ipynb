```python
def get_prompt(transcript = "Что-нибудть про LLM"):

    template = {
        "title": transcript.split('.')[0],
        "sections": [
            "План занятия с тайм кодами",
            "Основные концепции в формате Название - определение",
            "Ключевые персоны c указанием годов жизни в скобках и описанием исторического вклада. Проверяй фамилии и имена, так как при транскрибации они могут быть неточно распознаны",
            "Сравнение концепций c описанием идей в виде таблицы markdown",
            "Даты ключевых событий",
            "Чек лист для проверки из 5 вопросов с 3-мя вариантами ответов"
        ],
        "style": "Сделай краткий конспект по учебному стилю, как в ЦИФРИУМ или Лекториум. Добавь таймкоды, ключевые понятия и списки. Добавь ссылки на Википедию"
    }
    
    # Формируем prompt
    system_prompt = f"""Ты — помощник лектора, готовишь конспект. Преобразуй данный текст в структурированный учебный конспект в формате markdown
    Структура конспекта:
    - {template["title"]}
    - Разделы: {', '.join(template["sections"])}
    - Стиль: {template["style"]}
    
    Ответ начни сразу с заголовка и структурированного текста:
    """
    prompt = [
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": transcript}
    ]

    return prompt

js = get_prompt(transcript)
print(json.dumps(js,indent=4, ensure_ascii=False))
```

```python
tm()
from openai import OpenAI
from pathlib import Path

API_MODEL="o1-pro-2025-03-19"
API_MODEL="gpt-4"
API_MODEL="gpt-3.5-turbo"

def transform_transcript_to_conspect(api_key: str, input_file: str, api_model: str = API_MODEL):
    # Чтение исходного текста
    transcript = Path(input_file).read_text(encoding='utf-8')
    
    # Инициализация клиента (без proxies)
    client = OpenAI(api_key=api_key)
    
    
    # Запрос к модели
    response = client.chat.completions.create(
        model=api_model,  # Или "gpt-4", если доступен
        messages=get_prompt(transcript),
        temperature=0.5
    )
    
    # Сохранение результата
    result_text = response.choices[0].message.content
    # Path(output_file).write_text(conspect, encoding='utf-8')

    _model = api_model.replace("/", "_")
    output_file = f"conspects/conspect_{_model}.md"
    
    with open(output_file, "w", encoding="utf-8") as f:
        f.write(result_text.strip())
    
    print(f"✅ Конспект сохранен в {output_file}")

# Пример использования
if __name__ == "__main__":
    # API_KEY = "ваш_api_ключ"  # Замените на реальный ключ
    transform_transcript_to_conspect(
        api_key=KEY.OPENAI_API_KEY,
        input_file="transcript.txt",
        api_model=API_MODEL
    )

tm('>>>')
```

```python
tm()
import requests
import json
import os

# Endpoint API
url = "https://api.anthropic.com/v1/messages"
# Заголовки запроса

headers = {
    "x-api-key": KEY.ANTHROPIC_API_KEY, # API ключ Anthropic (рекомендуется хранить в переменных окружения)
    "anthropic-version": "2023-06-01",
    "content-type": "application/json"
}

fname = "transcript.txt"
with open(fname, "r", encoding="utf-8") as f:
    transcript = f.read()


# Шаблон для генерации
template = {
    "title": "Общественное движение в России во второй четверти XIX века",
    "sections": [
        "План занятия",
        "Основные концепции",
        "Ключевые персоны c указанием годов жизни",
        "Сравнение концепций c описанием идей в виде таблицы markdown",
        "Даты ключевых событий"
    ],
    "style": "Сделай краткий конспект. Добавь таймкоды, ключевые понятия. Избегай повторений."
}

prompt = f"""Ты — помощник-конспектолог. Преобразуй данный текст в структурированный учебный конспект. В формате учебных курсов Цифри 

Исходный текст:
\"\"\"
{transcript}
\"\"\"

Структура конспекта:
- Название: {template["title"]}
- Разделы: {', '.join(template["sections"])}
- Стиль: {template["style"]}

Ответ начни сразу с заголовка, продолжи в виде структурированного текста в формате markdown.

"""


# Данные запроса
data = {
    "model": "claude-3-7-sonnet-20250219",  # Используем актуальную модель
    "max_tokens": 2000,
    "messages": [
            {"role": "user", "content": prompt}
        ]
}

# Отправка запроса
response = requests.post(url, headers=headers, json=data)

# Проверка и вывод результата
if response.status_code == 200:
    result = response.json()
    tm("🧾Ответ от Claude:")
    conspect = result["content"][0]["text"]
    print(conspect)
    output_file = r'conspects/conspect_API_claude-3-7-sonnet-20250219.txt'
    Path(output_file).write_text(conspect, encoding='utf-8')
    
else:
    print(f"Ошибка: {response.status_code}")
    print(response.text)

tm('>>>')
```
