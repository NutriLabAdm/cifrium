```python
import os
import subprocess
import time

def print_time(txt):
    if not hasattr(print_time, "start_time"):
        print_time.start_time = time.time()
        print_time.lap_time = print_time.start_time

    curr_time = time.time()
    print(f"{curr_time - print_time.lap_time:10.3f} {curr_time - print_time.start_time:10.3f} {txt}")
    print_time.lap_time = curr_time

def get_video_duration(video_path):
    """ Получает длительность видео с использованием FFprobe. """
    command = ["ffprobe", "-v", "error", "-show_entries", "format=duration", "-of", "csv=p=0", video_path]
    result = subprocess.run(command, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)
    try:
        return float(result.stdout.strip())
    except ValueError:
        print("❌ Ошибка: невозможно определить длительность видео.")
        return 0

def split_video_with_ffmpeg(video_path, segment_duration, resize_percent=None):
    """ Разрезает видео на части с изменением размера в процентах (если указан). """
    if not os.path.exists(video_path):
        print(f"❌ Ошибка: файл {video_path} не найден.")
        return
    
    filename, ext = os.path.splitext(os.path.basename(video_path))
    total_duration = get_video_duration(video_path)
    if total_duration == 0:
        return
    
    print(f"🎥 Длительность видео: {total_duration:.2f} сек.")
    num_segments = int(total_duration // segment_duration) + (1 if total_duration % segment_duration > 0 else 0)
    print(f"🔪 Разрезаем на {num_segments} фрагментов...")

    output_folder = os.path.join(os.path.dirname(os.path.abspath(video_path)), f"{filename}_parts")
    os.makedirs(output_folder, exist_ok=True)

    for i in range(num_segments):
        start_time = i * segment_duration
        output_file = os.path.join(output_folder, f"{filename}_part_{i+1:02d}{ext}")

        command = [
            "ffmpeg", "-i", video_path,  # Исходное видео
            "-ss", str(start_time),  # Начало фрагмента
            "-t", str(segment_duration),  # Длительность
            "-c:v", "libx264", 
            "-c:a", "aac", 
            "-strict", "experimental"  # Должен идти перед аудиокодеком
        ]
        
        # Добавляем изменение размера, если указано
        if resize_percent:
            scale_filter = f"scale=iw*{resize_percent/100}:ih*{resize_percent/100}"
            command.extend(["-vf", scale_filter])
        
        command.append(output_file)  # Добавляем путь сохранения
        
        try:
            subprocess.run(command, check=True)
            print_time(f"✂️ {i+1}/{num_segments}: {start_time:.2f}-{start_time + segment_duration:.2f} сек → {output_file}")
        except subprocess.CalledProcessError as e:
            print(f"❌ Ошибка при обработке {output_file}: {e}")

    print(f"✅ Разделение завершено! Файлы сохранены в {output_folder}")

video_path = rf"лекции_2024_2025/Общественное движение в России 2четверь XIX в_01.mp4"  # Используй r"" для корректной обработки путей
segment_duration = 90
resize_percent = 50  # Уменьшение размера видео до 50%
    
split_video_with_ffmpeg(video_path, segment_duration, resize_percent)
print(f"💯 Файлы сохранены!")
```

```python
tm()

import os
import subprocess
import time
from PIL import Image

import cv2
import pytesseract
import numpy as np

pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'  # ← настрой под свою систему

def print_time(txt):
    if not hasattr(print_time, "start_time"):
        print_time.start_time = time.time()
        print_time.lap_time = print_time.start_time

    curr_time = time.time()
    print(f"{curr_time - print_time.lap_time:10.3f} {curr_time - print_time.start_time:10.3f} {txt}")
    print_time.lap_time = curr_time

def get_video_duration(video_path):
    command = ["ffprobe", "-v", "error", "-show_entries", "format=duration", "-of", "csv=p=0", video_path]
    result = subprocess.run(command, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)
    try:
        return float(result.stdout.strip())
    except ValueError:
        print("❌ Ошибка: невозможно определить длительность видео.")
        return 0

def extract_frame(video_path, time_seconds, output_image):
    command = [
        "ffmpeg", "-ss", str(time_seconds), "-i", video_path,
        "-frames:v", "1", "-q:v", "2", output_image,
        "-y"  # overwrite
    ]
    subprocess.run(command, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

def ocr_image(image_path):
    image = Image.open(image_path)
    return pytesseract.image_to_string(image, lang="rus+eng")


def ocr_in_region_2(image_path, region=None, min_width=50, min_height=20):
    # Загрузка изображения
    image = cv2.imread(image_path)
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Бинаризация
    _, thresh = cv2.threshold(gray, 150, 255, cv2.THRESH_BINARY_INV)

    # Если задана область для распознавания, вырезаем её
    if region:
        x1, y1, x2, y2 = region
        image = image[y1:y2, x1:x2]
        thresh = thresh[y1:y2, x1:x2]

    # Поиск контуров
    contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    results = []

    for cnt in contours:
        x, y, w, h = cv2.boundingRect(cnt)

        # Пропускаем слишком мелкие блоки
        if w < min_width or h < min_height:
            continue

        roi = image[y:y+h, x:x+w]

        # Конфигурация для распознавания одного абзаца
        custom_config = r'--psm 6'  # Treat the block as a single paragraph of text
        text = pytesseract.image_to_string(roi, config=custom_config)

        if text.strip():  # Если есть текст
            results.append({
                'box': (x, y, w, h),
                'text': text.strip()
            })
            # Нарисовать рамку вокруг блока
            cv2.rectangle(image, (x, y), (x+w, y+h), (0, 0, 255), 2)  # Красная рамка

            # Отобразить распознанный текст
            cv2.putText(image, text.strip(), (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (36,255,12), 1)

    # Отобразим изображение с рамкой
    plt.figure(figsize=(10, 10))
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.axis('off')  # Отключаем оси
    plt.show()

    return results
    

def ocr_in_region(image_path, region=None, min_width=50, min_height=20):
    """Распознавание текста в изображении с проверкой загрузки"""
    image = cv2.imread(image_path)
    
    if image is None:
        print(f"⚠️ Не удалось загрузить изображение: {image_path}")
        return ""

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    _, thresh = cv2.threshold(gray, 150, 255, cv2.THRESH_BINARY_INV)

    contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    texts = []

    for cnt in contours:
        x, y, w, h = cv2.boundingRect(cnt)
        if w > min_width and h > min_height:
            roi = image[y:y+h, x:x+w]
            text = pytesseract.image_to_string(roi, lang="rus+eng")
            texts.append(text.strip())

    return "\n".join(t for t in texts if t)


def split_video_with_ffmpeg(video_path, segment_duration, resize_percent=None):
    if not os.path.exists(video_path):
        print(f"❌ Ошибка: файл {video_path} не найден.")
        return

    filename, ext = os.path.splitext(os.path.basename(video_path))
    total_duration = get_video_duration(video_path)
    if total_duration == 0:
        return

    print(f"🎥 Длительность видео: {total_duration:.2f} сек.")
    num_segments = int(total_duration // segment_duration) + (1 if total_duration % segment_duration > 0 else 0)
    print(f"🔪 Разрезаем на {num_segments} фрагментов...")

    base_dir = os.path.dirname(os.path.abspath(video_path))
    output_folder = os.path.join(base_dir, f"parts")
    screenshots_folder = os.path.join(base_dir, f"screenshots")
    os.makedirs(output_folder, exist_ok=True)
    os.makedirs(screenshots_folder, exist_ok=True)

    ocr_output_path = os.path.join(base_dir, f"{filename}_ocr.txt")
    
    with open(ocr_output_path, "w", encoding="utf-8") as ocr_file:
        for i in range(num_segments):
            start_time = i * segment_duration
            output_file = os.path.join(output_folder, f"{filename}_part_{i+1:02d}{ext}")

            command = [
                "ffmpeg", "-i", video_path,
                "-ss", str(start_time),
                "-t", str(segment_duration),
                "-c:v", "libx264",
                "-c:a", "aac",
                "-strict", "experimental"
            ]

            if resize_percent:
                scale_filter = f"scale=iw*{resize_percent/100}:ih*{resize_percent/100}"
                command.extend(["-vf", scale_filter])

            command.append(output_file)

            try:
                subprocess.run(command, check=True)
                print_time(f"✂️ {i+1}/{num_segments}: {start_time:.2f}-{start_time + segment_duration:.2f} сек → {output_file}")

                # === Скриншот ===
                image_file = os.path.join(screenshots_folder, f"{filename}_frame_{i+1:02d}.jpg")
                extract_frame(output_file, 2, image_file)  # снимок на 2 секунде
                print_time(f"📸 Скриншот: {image_file}")

                # === OCR ===
                text = ocr_in_region(image_file).strip()
                
                if text:
                    ocr_file.write(f"## Фрагмент {i+1}\n{text}\n\n")
                    print_time(f"🔤 OCR найден текст ({len(text)} символов)")

            except subprocess.CalledProcessError as e:
                print(f"❌ Ошибка при обработке {output_file}: {e}")

    print(f"\n✅ Разделение завершено! Файлы сохранены в {output_folder}")
    print(f"📝 Текст из изображений сохранен в: {ocr_output_path}")

# === Запуск ===
video_path = r"lectures_2025/Общественное движение в России 2четверь XIX в_01.mp4"
segment_duration = 45
resize_percent = 75

split_video_with_ffmpeg(video_path, segment_duration, resize_percent)

tm('>>>')
```
