```python
import os
import subprocess
import time

def print_time(txt):
    if not hasattr(print_time, "start_time"):
        print_time.start_time = time.time()
        print_time.lap_time = print_time.start_time

    curr_time = time.time()
    print(f"{curr_time - print_time.lap_time:10.3f} {curr_time - print_time.start_time:10.3f} {txt}")
    print_time.lap_time = curr_time

def get_video_duration(video_path):
    """ Получает длительность видео с использованием FFprobe. """
    command = ["ffprobe", "-v", "error", "-show_entries", "format=duration", "-of", "csv=p=0", video_path]
    result = subprocess.run(command, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)
    try:
        return float(result.stdout.strip())
    except ValueError:
        print("❌ Ошибка: невозможно определить длительность видео.")
        return 0

def split_video_with_ffmpeg(video_path, segment_duration, resize_percent=None):
    """ Разрезает видео на части с изменением размера в процентах (если указан). """
    if not os.path.exists(video_path):
        print(f"❌ Ошибка: файл {video_path} не найден.")
        return
    
    filename, ext = os.path.splitext(os.path.basename(video_path))
    total_duration = get_video_duration(video_path)
    if total_duration == 0:
        return
    
    print(f"🎥 Длительность видео: {total_duration:.2f} сек.")
    num_segments = int(total_duration // segment_duration) + (1 if total_duration % segment_duration > 0 else 0)
    print(f"🔪 Разрезаем на {num_segments} фрагментов...")

    output_folder = os.path.join(os.path.dirname(os.path.abspath(video_path)), f"{filename}_parts")
    os.makedirs(output_folder, exist_ok=True)

    for i in range(num_segments):
        start_time = i * segment_duration
        output_file = os.path.join(output_folder, f"{filename}_part_{i+1:02d}{ext}")

        command = [
            "ffmpeg", "-i", video_path,  # Исходное видео
            "-ss", str(start_time),  # Начало фрагмента
            "-t", str(segment_duration),  # Длительность
            "-c:v", "libx264", 
            "-c:a", "aac", 
            "-strict", "experimental"  # Должен идти перед аудиокодеком
        ]
        
        # Добавляем изменение размера, если указано
        if resize_percent:
            scale_filter = f"scale=iw*{resize_percent/100}:ih*{resize_percent/100}"
            command.extend(["-vf", scale_filter])
        
        command.append(output_file)  # Добавляем путь сохранения
        
        try:
            subprocess.run(command, check=True)
            print_time(f"✂️ {i+1}/{num_segments}: {start_time:.2f}-{start_time + segment_duration:.2f} сек → {output_file}")
        except subprocess.CalledProcessError as e:
            print(f"❌ Ошибка при обработке {output_file}: {e}")

    print(f"✅ Разделение завершено! Файлы сохранены в {output_folder}")

video_path = rf"лекции_2024_2025/Общественное движение в России 2четверь XIX в_01.mp4"  # Используй r"" для корректной обработки путей
segment_duration = 90
resize_percent = 50  # Уменьшение размера видео до 50%
    
split_video_with_ffmpeg(video_path, segment_duration, resize_percent)
print(f"💯 Файлы сохранены!")
```
