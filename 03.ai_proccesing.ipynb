### 04. openAI  transform_transcript_to_conspect

```python
tm()
from openai import OpenAI
from pathlib import Path

API_MODEL="o1-pro-2025-03-19"
API_MODEL="gpt-4"
API_MODEL="gpt-3.5-turbo"

def transform_transcript_to_conspect(api_key: str, input_file: str, output_file: str):
    # Чтение исходного текста
    transcript = Path(input_file).read_text(encoding='utf-8')
    
    # Инициализация клиента (без proxies)
    client = OpenAI(api_key=api_key)
    

    # Шаблон (можно вынести в JSON для кастомизации)
    template = {
        "title": "Общественное движение в России во второй четверти XIX века",
        "sections": [
            "План занятия",
            "Основные концепции",
            "Ключевые персоны",
            "Сравнение концепций в таблице",
            "Даты ключевых событий"
        ],
        "style": "Сделай краткий конспект по учебному стилю, как в ЦИФРИУМ или Лекториум. Добавь таймкоды, ключевые понятия и списки."
    }
    
    # Формируем prompt
    system_prompt = f"""Ты -- помощник-конспектолог. Преобразуй данный текст в структурированный учебный конспект.
    
    Исходный текст:
    \"\"\"
    {transcript[:4096]}
    \"\"\"
    
    Структура конспекта:
    - Название: {template["title"]}
    - Разделы: {', '.join(template["sections"])}
    - Стиль: {template["style"]}
    
    Ответ начни сразу с заголовка и структурированного текста:
    """

    
    
    # Запрос к модели
    response = client.chat.completions.create(
        model=API_MODEL,  # Или "gpt-4", если доступен
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": transcript}
        ],
        temperature=0.3
    )
    
    # Сохранение результата
    conspect = response.choices[0].message.content
    Path(output_file).write_text(conspect, encoding='utf-8')
    print(f"✅ Конспект сохранен в {output_file}")

# Пример использования
if __name__ == "__main__":
    # API_KEY = "ваш_api_ключ"  # Замените на реальный ключ
    transform_transcript_to_conspect(
        api_key=KEY.OPENAI_API_KEY,
        input_file="transcript.txt",
        output_file=f"conspect_{MODEL}.txt"
    )

tm('>>>')
```
