### 01.01 simple pptx

```python
from pptx import Presentation
from pptx.util import Inches, Pt
import re

tm()

def parse_text_file(filename):
    with open(filename, encoding='utf-8') as f:
        lines = f.readlines()
    # Извлечение заголовка и подзаголовка
    title = lines[0].strip()
    subtitle = lines[1].strip()
    content = ''.join(lines[2:])
    # sections = re.split(r'# |\n', content)

    pattern = r'#\s*(.+?)\s*(?=\n|$)(.*?)(?=(#\s|$))'
    
    
    # Извлечение заголовков и содержания
    sections = re.findall(pattern, content, re.DOTALL)
    # print (sections)
    
        
    blocks = {}

    for section in sections:
        if len(section) == 3:
            header,body, _ = section
            blocks[header.strip()] = body.strip()
        else:
            blocks[header.strip()] += '\n' + section.strip()

    return title, subtitle, blocks

def create_pptx(title, subtitle, blocks, template_file, output_file):
    prs = Presentation(template_file)

    # Добавление титульного слайда
    title_slide_layout = prs.slide_layouts[0]  # Макет титульного слайда
    slide = prs.slides.add_slide(title_slide_layout)
    slide.shapes.title.text = title
    slide.placeholders[1].text = subtitle

    for section, content in blocks.items():
        slide_layout = prs.slide_layouts[1]  # Макет слайда с заголовком и содержимым
        slide = prs.slides.add_slide(slide_layout)
        slide.shapes.title.text = section

        content_placeholder = slide.placeholders[1]
        content_placeholder.text = content

    prs.save(output_file)
    print(f"✅ Презентация сохранена в: {output_file}")


STRUCTURIZED_FILE = r'conspects\\conspect_API_claude-3-7-sonnet-20250219.md'
TEMPLATE_FILE = 'themes.pptx'

output_file = rf'pptx/{STRUCTURIZED_FILE[10:-4]}.pptx'

if __name__ == "__main__":
    title, subtitle, blocks = parse_text_file(STRUCTURIZED_FILE)
    create_pptx(title, subtitle, blocks, TEMPLATE_FILE, output_file)
tm('>>>')
```

### 01.03 conspect to pptx with table

```python
tm()

from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.dml.color import RGBColor
from pptx.enum.shapes import MSO_AUTO_SHAPE_TYPE
from bs4 import BeautifulSoup
import markdown

# === Параметры ===
CONSPECT_FILE = r'conspects\\conspect_API_claude-3-7-sonnet-20250219.md'
CONSPECT_FILE = r'conspects\\conspect_models_meta-llama-3-8b-instruct.Q4_K_M.gguf.md'
CONSPECT_FILE = r'conspects\\conspect_models_claude-3.7-sonnet-reasoning-gemma3-12B.Q8_0.gguf.md'
CONSPECT_FILE = r'conspects/conspect_claude-3.7-sonnet-reasoning-gemma3-12B.Q8_0.gguf.md'
# CONSPECT_FILE = r'conspects\\conspect_gpt-3.5-turbo.md'


TEMPLATE_FILE = 'themes.pptx'
OUTPUT_FILE = f"{CONSPECT_FILE.replace('conspects','pptx')}.pptx"

FONT_SIZE = Pt(9)

# === Утилиты ===
def set_cell_font_size(cell, font_size=Pt(10)):
    for paragraph in cell.text_frame.paragraphs:
        for run in paragraph.runs:
            run.font.size = font_size

def set_run_style(run, bold=False, color = RGBColor(0, 0 , 0), italic=False, font_size=Pt(10)):
    run.font.bold = bold
    run.font.italic = italic
    run.font.size = font_size
    # run.font.name = 'Arial'
    run.font.color.rgb = color

def parse_html_to_slide(title, html_content, slide, font_size=Pt(10)):
    textbox = slide.placeholders[1]
    textbox.text = ""
    tf = textbox.text_frame

    soup = BeautifulSoup(html_content, "html.parser")

    for element in soup.contents:
        if not hasattr(element, "name"):
            continue

        if element.name == "p":
            p = tf.add_paragraph()
            for part in element.descendants:
                if part.name is None:
                    run = p.add_run()
                    run.text = part
                    # set_run_style(run)
                elif part.name == "strong_":
                    run = p.add_run()
                    run.text = part.text
                    set_run_style(run, bold=True)
                elif part.name == "em_":
                    run = p.add_run()
                    run.text = part.text
                    set_run_style(run, italic=True)

        elif element.name in ["h3", "h4", "h5"]:
            p = tf.add_paragraph()
            run = p.add_run()
            run.text = element.text
            set_run_style(run, bold=True, font_size=Pt(20), color=RGBColor(220, 110 , 240))
            p.level = 0

        elif element.name == "ul":
            for li in element.find_all("li"):
                p = tf.add_paragraph()
                p.text = f"• {li.text}"
                p.level = 0

        elif element.name == "ol":
            for idx, li in enumerate(element.find_all("li"), 1):
                p = tf.add_paragraph()
                p.text = f"{idx}. {li.text}"
                p.level = 0

        elif element.name == "table":
            rows = element.find_all("tr")
            cols_count = len(rows[0].find_all(["th", "td"]))
            table_shape = slide.shapes.add_table(len(rows), cols_count, Inches(1), Inches(2.5), Inches(10), Inches(0.7 * len(rows)))
            table = table_shape.table
            for r, row in enumerate(rows):
                for c, cell in enumerate(row.find_all(["td", "th"])):
                    table.cell(r, c).text = cell.text
                    set_cell_font_size(table.cell(r, c), font_size)

# === Загрузка и подготовка Markdown ===
with open(CONSPECT_FILE, encoding='utf-8') as f:
    markdown_text = f.read()

html = markdown.markdown(markdown_text, extensions=['tables', 'extra'])
soup = BeautifulSoup(html, "html.parser")

prs = Presentation(TEMPLATE_FILE)
title_slide_layout = prs.slide_layouts[0]
bullet_slide_layout = prs.slide_layouts[1]

# === Управление слайдами ===
def add_title_slide(title):
    slide = prs.slides.add_slide(title_slide_layout)
    slide.shapes.title.text = title

def add_slide(title, html_content, font_size=Pt(10)):
    slide = prs.slides.add_slide(bullet_slide_layout)
    slide.shapes.title.text = title
    parse_html_to_slide(title, html_content, slide, font_size)

# === Парсинг ===
current_title = None
current_html_content = []
first_slide_added = False

for element in soup.children:
    if not hasattr(element, "name"):
        continue

    if element.name == "h1" and not first_slide_added:
        add_title_slide(element.text)
        first_slide_added = True

    elif element.name == "h2":
        if current_title:
            add_slide(current_title, "".join(str(e) for e in current_html_content), FONT_SIZE)
            current_html_content = []
        current_title = element.text
    else:
        current_html_content.append(element)

if current_title and current_html_content:
    add_slide(current_title, "".join(str(e) for e in current_html_content), FONT_SIZE)

prs.save(OUTPUT_FILE)
tm(f"Готово! >>> {OUTPUT_FILE} ")
```
